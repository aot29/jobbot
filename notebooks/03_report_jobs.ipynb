Load all job assessments

```python
import pandas as pd
from pathlib import Path
```

## Read the jobs

```python
folder = Path("../jobs")  # Directory containing the jobs
```

```python
df_jobs = pd.concat(
    [pd.read_excel(f) for f in folder.glob("*.xlsx") if not f.name.startswith("~$")],
    ignore_index=True
)
```

```python
df_jobs.head()
```

## Read the assessments

```python
folder = Path("../assessments")  # Directory containing the assessments
```

```python
df_assessments = pd.concat(
    [pd.read_csv(f) for f in folder.glob("*.csv") if not f.name.startswith("~$")],
    ignore_index=True
)
```

```python
df_assessments.head()
```

## Compute overal score for each job

```python
overal_score = []
for i,row in df_jobs.iterrows():
    job_assessments = df_assessments[df_assessments.JobID==row.id]
    if len(job_assessments) > 0:
        overal_score.append(job_assessments.Score.sum()/len(job_assessments))
    else:
        overal_score.append(None)

df_jobs['score'] = overal_score
```

## Display fitting jobs
Remember that 1. Perfect fit, 2. Good Fit, 3. Probable Fit, 4. Poor Fit, 5. Not a Fit.

```python
# Drop rows where score is None/NaN
valid = df_jobs.dropna(subset=["score"])

# 1st percentile = the threshold below which the best 1% of scores fall
cutoff = valid["score"].quantile(0.01)

best = valid[valid["score"] <= cutoff]
best = best.sort_values("score")

print(f"Found {len(best)} jobs in the 1% best scores fitting the profile")
for i, row in best.iterrows():
    print("-" * 120)
    print(f"{i}: Score: {row.score}, posted on:{row.posted}")
    print(f"Company: {row.company}, Location: {row.location}")
    print(row.title)
    print(row.description)
    print(row.url)
    
```
